```python
def fetch_lyrics_in_batches(api, df, track_col='track_name', artist_col='artist_name', batch_size=100):
    """
    Fetch lyrics for songs in batches from Lyrics Genius API and append them to a DataFrame.

    Arguments:
    - api: An instance of a lyrics API client that provides the `search_song(title, artist)` method.
           This API client is expected to fetch lyrics based on track and artist names.
    - df: A pandas DataFrame containing at least two columns: one for track names and one for artist names.
    - track_col: (str, optional) The name of the column in the DataFrame that contains the track names.
                 Default is 'track_name'.
    - artist_col: (str, optional) The name of the column in the DataFrame that contains the artist names.
                  Default is 'artist_name'.
    - batch_size: (int, optional) The number of tracks to process in each batch. Default is 100.

    Output:
    - Returns a pandas DataFrame which is the original DataFrame with an added 'lyrics' column.
      This column contains the lyrics for each track if found, otherwise an appropriate message like
      'Lyrics not found' or an error message if an issue occurred during the fetch operation.
    """
    # Initialize an empty list to store all the lyrics
    all_lyrics = []

    # Extract track names and artist names as lists from the DataFrame
    tracks = df[track_col].tolist()
    artists = df[artist_col].tolist()

    # Loop through the track and artist lists in batches, with each batch
    for i in range(0, len(tracks), batch_size):
        # Create a batch of track names and a corresponding batch of artist names
        batch_tracks = tracks[i:i + batch_size]
        batch_artists = artists[i:i + batch_size]
        
        # Initialize a list to store lyrics for the current batch
        batch_lyrics = []

        # Iterate over each track and artist pair in the batch
        for track, artist in zip(batch_tracks, batch_artists):
            try:
                # Use the API to search for the song by title and artist
                song = api.search_song(title=track, artist=artist)
                
                # If the song is found, get the lyrics, otherwise append 'Lyrics not found'
                if song:
                    lyric = song.lyrics
                    batch_lyrics.append(lyric)
                else:
                    batch_lyrics.append('Lyrics not found')
            except Exception as e:
                # Handle any exception that occurs during the API call
                # Print the error and append an error message to the lyrics list
                print(f'Error fetching lyrics for {track} by {artist}: {str(e)}')
                batch_lyrics.append(f'Error: {str(e)}')

        # Add the lyrics from the current batch to the master list of all lyrics
        all_lyrics.extend(batch_lyrics)

    # Append the collected lyrics as a new column in the DataFrame
    df['lyrics'] = all_lyrics

    # Return the updated DataFrame with the lyrics included
    return df

fetch_lyrics_in_batches(api, gym)
```

# Musical Features from Spotify

```python
# Songs in each album using Spotify

SPOTIPY_CLIENT_ID='YOUR_SPOTIFY_CLIENT_ID'
SPOTIPY_CLIENT_SECRET='YOUR_SPOTIFY_CLIENT_SECRET'

# Initialize Spotify client credentials manager
auth_manager = SpotifyClientCredentials(client_id=SPOTIPY_CLIENT_ID, client_secret=SPOTIPY_CLIENT_SECRET)

# Create a Spotify client
sp = spotipy.Spotify(auth_manager=auth_manager)
```

```python
import time
import pandas as pd

def get_audio_features(sp, df, track_id_col='track_id', batch_size=100):
    """
    Fetch audio features for tracks in batches using Spotify's API client and append them to a DataFrame.

    Arguments:
    - sp: A Spotify API client (e.g., Spotipy instance) that provides the `audio_features()` method
          to retrieve audio features based on track IDs.
    - df: A pandas DataFrame containing a column of Spotify track IDs.
    - track_id_col: (str, optional) The name of the column in the DataFrame that contains track IDs.
                    Default is 'track_id'.
    - batch_size: (int, optional) The number of track IDs to process in each batch. Default is 100.

    Output:
    - Returns a pandas DataFrame which is the original DataFrame with additional columns representing
      audio features such as danceability, energy, loudness, tempo, etc.

    """

    # Initialize an empty list to store audio features for all tracks
    audio_features = []

    # Extract track IDs from the DataFrame
    track_ids = df[track_id_col].tolist()

    # Process track IDs in batches to handle API rate limits effectively
    for i in range(0, len(track_ids), batch_size):
        batch = track_ids[i:i + batch_size]

        # Handle rate limiting and errors with a retry mechanism
        while True:
            try:
                # Fetch audio features for the batch of track IDs using the Spotify API
                audio = sp.audio_features(batch)
                # Add the retrieved audio features to the list
                audio_features.extend(audio)
                break
            except spotipy.exceptions.SpotifyException as e:
                if e.http_status == 429:  # Handle rate-limiting errors (HTTP 429)
                    # Retrieve the 'Retry-After' header and wait accordingly
                    retry_after = int(e.headers.get('Retry-After', 1200))
                    print(f"Rate limited. Retrying after {retry_after} seconds...")
                    time.sleep(retry_after)
                else:
                    # Handle other types of errors by printing and breaking the loop
                    print(f"An error occurred: {e}")
                    break

    # Initialize a dictionary to store audio feature data, with each feature as a list
    data = {
        'danceability': [],
        'energy': [],
        'key': [],
        'loudness': [],
        'mode': [],
        'speechiness': [],
        'acousticness': [],
        'liveness': [],
        'valence': [],
        'tempo': [],
        'id': []
    }

    # Extract each feature from the audio features returned by the API
    for audio in audio_features:
        for key in data:
            data[key].append(audio[key])

    # Create a new DataFrame from the extracted audio feature data
    track_feature = pd.DataFrame(data)

    # Rename the 'id' column to 'Track ID' for merging purposes
    track_feature.rename(columns={'id': 'Track ID'}, inplace=True)

    # Merge the original DataFrame with the new audio features DataFrame using 'Track ID'
    result_df = df.join(track_feature.set_index('Track ID'), on=track_id_col)

    # Return the merged DataFrame
    return result_df
```
